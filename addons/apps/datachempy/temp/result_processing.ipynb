Import libraries

```python
import json
import os
import numpy as np
```

```python
# Print the current directory
print(os.getcwd())
```

```python
path = os.getcwd().split('/')
path = path[0:len(path)-4]
path = '/'.join(path)
file = "/result.json"
path = path + file

# Print the path
print(path)
```

```python
# Read the JSON file

with open(path, 'r') as file:
    data = json.load(file)

# Print the keys of the dictionary
print(data.keys())
```

```python
def remove_extras(input_string):
    """
    Detects the presence of '±' in the input string and removes everything after it.
    
    Parameters:
        input_string (str): The string to process.
    
    Returns:
        str: The processed string with content after '±' removed.
    """
    if "±" in input_string:
        return input_string.split("±")[0].strip()
    return input_string
```

```python
for name in data.keys():
    for id in data[name].keys():
        for section in data[name][id].keys():
            if section == "name":
                print("Name: ", data[name][id][section])
            if section == "ID":
                print("ID: ", data[name][id][section])
            if section == "data_table":
                for ref in data[name][id]["data_table"].keys():
                    if ref == "cTP":
                        for table in data[name][id]["data_table"][ref].keys():
                            if table == "One dimensional data":
                                print("One dimensional data: ", data[name][id]["data_table"][ref][table][1][0:2])
```

```python
parameter = "Tfus"
```

```python
min = 0
max = 300
```

```python

# Iterate through each compound in the data
for compound_name, compound_data in data.items():
    for compound_id, compound_details in compound_data.items():
        if "data_table" in compound_details:
            for key, subkey in compound_details["data_table"].items():
                for table, rows in compound_details["data_table"][key].items():
                    parameter_value = np.array([])
                    for row in rows:
                        if (
                            isinstance(row, list)
                            and parameter in row 
                            and len(row) > 2
                            ):
                            parameter_value = np.append(parameter_value, float(remove_extras(row[1])))
                    if (
                        parameter_value.size > 0
                        and parameter_value.mean() > min
                        and parameter_value.mean() < max
                        ):
                        print (f"Compound: {compound_name}, ID: {compound_id}")
                        print(f"{parameter}: {parameter_value.mean()}")
                            
```

```python

values = [float(remove_extras(row[1])) for compound_name, compound_data in data.items() for compound_id, compound_details in compound_data.items() if "data_table" in compound_details for key, subkey in compound_details["data_table"].items() for table, rows in subkey.items() for row in rows if isinstance(row, list) and parameter in row and len(row) > 2]
mean_value = np.mean(values)
print("Mean value:", mean_value)
```
