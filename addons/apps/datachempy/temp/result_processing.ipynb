```python
def remove_extras(input_string):
    """
    Detects the presence of '±' in the input string and removes everything after it.
    
    Parameters:
        input_string (str): The string to process.
    
    Returns:
        str: The processed string with content after '±' removed.
    """
    if "±" in input_string and "to" in input_string:
        return input_string.split("±")[0].strip()
    if "±" in input_string and "to" not in input_string:
        return input_string.split("±")[0].strip()
    if "to" in input_string and "±" not in input_string:
        return input_string.split("to")[0].strip()
    
    if "to" in input_string:
        return input_string.split("to")[0].strip()
    if "±" in input_string:
        return input_string.split("±")[0].strip()
    
    return input_string
```

```python
parameter_list = ["Tfus","Tboil","Pc","Tc","ρc","\u0394fH\u00b0gas","\u0394fH\u00b0liquid","\u0394fH\u00b0solid"]
compound_descriptor = ["name","synonyms","formula","mol_weight","inchi","inchi_key","cas_rn","mol_refs"]
```

```python
# Iterate through each compound in the data
compound_dict = {}
for compound_name, compound_data in data.items():
    for compound_id, compound_details in compound_data.items():
        compound_dict[compound_id] = {}
        compound_dict[compound_id]["search_name"] = compound_name
        for detail, description in compound_details.items():
            if detail in compound_descriptor:
                compound_dict[compound_id][detail] = description
            #print(compound_dict)
        if "data_table" in compound_details:
            for parameter in parameter_list:
                parameter_value = np.array([])
                for key, subkey in compound_details["data_table"].items():
                    for table, rows in compound_details["data_table"][key].items():
                        # Create a dictionary to store the values for the compound
                        compound_dict[compound_id][parameter] = {}
                        
                        for row in rows:
                            if (
                                True
                                and isinstance(row, list)
                                and parameter in row 
                                and len(row) > 2
                                ):
                                #print(remove_extras(row[1]))
                                #print(row[1])
                                if remove_extras(row[1]) != "":
                                    value = float(remove_extras(row[1]))
                                    parameter_value = np.append(parameter_value, value)
                compound_dict[compound_id][parameter] = parameter_value.mean().round(2) if parameter_value.size > 0 else None
                


#print(compound_dict)
# Create a DataFrame from the dictionary
df = pd.DataFrame.from_dict(compound_dict, orient='index')
# Reset the index to have a default integer index
df.reset_index(inplace=True)
# Rename the index column
df.rename(columns={'index': 'ID'}, inplace=True)
# Print the DataFrame
#df['num_carbons'] = df['formula'].str.extract(r'C(\d+)')[0].astype(float)
# Handle cases where 'C' is present without a number or not present at all
df['num_carbons'] = df['formula'].str.extract(r'C(\d*)')[0].replace('', 1).fillna(0).astype(float)
df['num_carbons'] = df['num_carbons'].astype(int)
# Move the 'num_carbons' column next to 'formula'
columns = df.columns.tolist()
columns.insert(columns.index('formula') + 1, columns.pop(columns.index('num_carbons')))
df = df[columns]
print(df)
```

```python
df.to_csv('results.csv', index=False, sep=';')
```
